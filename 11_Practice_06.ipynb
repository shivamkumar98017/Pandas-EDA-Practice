```python
import numpy as np
import pandas as pd

df = pd.DataFrame(
    {
        "Employee": ["Amit", "Neha", "Ravi", "Priya", "Karan"],
        "Department": ["IT", "HR", "IT", "HR", "IT"],
        "Performance_Score": [85, 92, 85, 95, 78],
        "Experience": [3, 5, 4, 6, 2],
    }
)

df
```

```python
# Sort employees by dipartment and performance score.
df.sort_values(by=["Department","Performance_Score"])
```

```python
# Sort employess by performance score in descending order.Find the top 3 performers.
dec_top3 = df.sort_values(by="Performance_Score",ascending=False).head(3)
dec_top3
```

```python
largest3 = df.nlargest(3,"Performance_Score")
largest3
```

```python
# Assign dense ranks based on performance score.
df["Rank"]= df["Performance_Score"].rank(
    ascending=False,
    method="dense"
)
```

```python
df
```

```python
# instead of 0,1,2,3... give the new indexing : "A","B" and so on...
df.index = ["A","B","C","D","E"]
df
```

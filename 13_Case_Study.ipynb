```python
import pandas as pd
import numpy as np

data = {
    'Order_ID': [1001, 1002, 1003, 1004, 1005, 1006,
                 1007, 1008, 1009, 1010, 1011, 1001],

    'Customer_Name': ['Amit', 'Priya', 'Rahul', 'Sneha',
                      'Karan', 'Neha', 'Rohit', 'Anjali',
                      'Vikas', 'Pooja', 'Arjun', 'Amit'],

    'Product': ['Laptop', 'Mobile', 'Headphones', 'Tablet',
                'Laptop', 'Mouse', 'Keyboard', 'Mobile',
                'Monitor', 'Tablet', 'Headphones', 'Laptop'],

    'Category': ['Electronics', 'electronics', 'Accessories',
                 'Electronics', 'Electronics', 'accessories',
                 'Accessories', 'Electronics', 'electronics',
                 'Electronics', 'Accessories', 'Electronics'],

    'Quantity': ['1', '2', '1', 'two', '1', '3',
                 '2', '1', '1', '2', '1', '1'],

    'Price': ['55000', '20000', '1500', '18000',
              '52000', '500', '1200', '22000',
              '15000', '17000', '1800', '55000'],

    'Order_Status': ['Delivered', 'delivered', 'Pending',
                     'CANCELLED', 'Delivered', 'pending',
                     'Delivered', 'Shipped', 'delivered',
                     'Pending', 'Delivered', 'Delivered'],

    'Rating': [5, 4, np.nan, 3, 5, 4,
               3, np.nan, 4, 5, 2, 5]
}

df = pd.DataFrame(data)

print(df)
```

```python
df
```

```python
# Display only Order_ID,Product, and Price.
df[["Order_ID", "Product", "Price"]]
```

```python
# Find orders where Price is greater than 20000.
df[(df["Price"]) > "20000"]
```

```python
# Find Orders belonging to Electronics.
df[df["Category"]=="Electronics"]
```

```python
# Find delivered orders having a rating of at least 4.
df[
    (df["Order_Status"]=="Delivered") &
    (df["Rating"]>=4)
    ]
```

```python
# Check the current data types.
type(df)
```

```python
df.dtypes
```

```python
# Convert Quantity into numeric formate.
df["Quantity"] = pd.to_numeric(
    df['Quantity'],
    errors='coerce'
)
```

```python
# Convert Price into numeric formate.
df["Price"] = pd.to_numeric(
    df['Price'],
    errors='coerce'
)
```

```python
# Identify records where Quantity could not be converted.
df[df["Quantity"].isnull()]
```

```python
# Repalce invalid Quantity values with the median Quantity.
df["Quantity"] = df['Quantity'].fillna(df["Quantity"].median())
```

```python
# Count missing Ratings.
df["Rating"].isnull().sum()
```

```python
# Calculate the percentage of missing ratings.
perc = df["Rating"].isnull().sum()*100/len(df["Rating"])
perc
```

```python
print(df["Rating"].isnull().mean()*100)
```

```python
# Repalce missing ratings with with the median rating.
df["Rating"] = df["Rating"].fillna(
    df["Rating"].median()
)
df["Rating"].isnull().sum()
```

```python
# Find the duplicate Order_ID records.
df[df.duplicated(subset=["Order_ID"],keep=False)]
```

```python
# Display all duplicate order IDs.
df[df.duplicated(subset=["Order_ID"], keep=False)]["Order_ID"]
```

```python
# Find the number of unique orders.
df["Order_ID"].unique()
```

```python
# Remove duplicate orders while retaining the first record.
df.drop_duplicates(subset="Order_ID", keep="first")
```

```python
# Find Uniqe category Values.
df["Category"].unique()
```

```python
# Convert Category Values to the title case.
df["Category"] = df["Category"].str.strip().str.title()
```

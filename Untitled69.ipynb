```python
import pandas as pd
products = pd.DataFrame({
"Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Webcam"],
"Category": ["Electronics", "Accessories", "Accessories",
"Electronics", "Accessories"],
"Price": [55000, 700, 1500, 12000, 2500],
"Stock": [10, 100, 50, 15, 25]
})
products
```

```python
# Products costing less than 2000
products[products["Price"]<2000]
```

```python
# Products with fewer than 20 units in Stock.
products[products["Stock"]< 20]
```

```python
# Electronics only
products[products["Category"]=="Electronics"]
```

```python
# Create a new column with the total_inventory amount for each product.
products["Total_amount"] = products["Price"]*products["Stock"]
products
```

```python
# Select Product and stock columns for low_stock products.
products.loc[
    products["Stock"]<20,
    ["Product","Stock"]
]
```

```python
# What is the percentage contribution of each product to the overall inventory amount?
products["precContri"] = round(products["Total_amount"]*100/sum(products["Total_amount"]),2)
```

```python
products
```

```python
import pandas as pd
import numpy as np

sales = pd.DataFrame({
    "Product": ["Laptop", "Mouse", "Monitor", "Keyboard", "Printer"],
    "Quantity": [5, 20, 4, 10, 3],
    "Unit_Price": [50000, 800, 12000, 1500, 18000],
    "Cost": [220000, 10000, 40000, 10000, 45000]
})
```

```python
sales
```

```python
# Calcualte revenue and profit for each product,
# then classify products as high profit if profit is at least 10000;
# otherwise,Standard profit.
```

```python
sales["Revenue"] = sales["Quantity"]*sales["Unit_Price"]
```

```python
sales
```

```python
sales["Profit"] = sales["Revenue"]-sales["Cost"]
```

```python
sales
```

```python
sales["Profit_Category"] = np.where(sales["Profit"]>= 10000,"High Profit","Standard Profit")
```

```python
sales
```

```python
students = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Karan"],
    "Python": [85, 95, 45, 70, 55],
    "SQL": [80, 90, 50, 65, 60],
    "Statistics": [90, 92, 40, 75, 50]
})
students
```

```python
# Problem:calculate total marks and percantage for three subjects.
# Assign grade A for a percentage of at least 80,
# B for at least 60 but below 80, and c otherwise.
```

```python
students["Total_Marks"] = students["Python"] + students["SQL"] + students["Statistics"]
```

```python
students
```

```python
students["Percentage"] = round((students["Total_Marks"]*100)/300,2)
```

```python
students
```

```python
students["Grade"] = np.where(students["Percentage"] >= 80, "A", np.where(students["Percentage"] >= 60, "B", "C"))
```

```python
students
```

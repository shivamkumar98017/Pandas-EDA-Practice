# 11. GroupBy and Aggregation in Pandas

This notebook covers one of the most powerful features of Pandas — **GroupBy**.

You will learn:
- How to group data using `groupby()`
- Common aggregation functions (`sum`, `mean`, `count`, `min`, `max`, `agg`)
- Multi-column grouping
- Custom aggregations
- Practical exercises

---

```python
import pandas as pd
import numpy as np

# Sample sales data
data = {
    'Date': pd.date_range('2024-01-01', periods=12, freq='D'),
    'Region': ['North', 'South', 'East', 'West', 'North', 'South',
               'East', 'West', 'North', 'South', 'East', 'West'],
    'Product': ['A', 'B', 'A', 'C', 'B', 'A', 'C', 'B', 'A', 'C', 'B', 'A'],
    'Sales': [250, 180, 320, 150, 210, 290, 175, 240, 310, 160, 200, 280],
    'Quantity': [10, 7, 12, 5, 8, 11, 6, 9, 13, 4, 7, 10],
    'Profit': [45, 30, 60, 25, 35, 55, 28, 40, 58, 22, 32, 50]
}

df = pd.DataFrame(data)
df
```

## 1. Basic GroupBy

Group by a single column and apply an aggregation.

```python
# Total sales by Region
df.groupby('Region')['Sales'].sum()
```

```python
# Average sales by Product
df.groupby('Product')['Sales'].mean().round(2)
```

```python
# Multiple aggregations on one column
df.groupby('Region')['Sales'].agg(['sum', 'mean', 'count', 'min', 'max'])
```

## 2. GroupBy on Multiple Columns

```python
# Sales by Region and Product
df.groupby(['Region', 'Product'])['Sales'].sum().unstack(fill_value=0)
```

```python
# More detailed multi-level aggregation
df.groupby(['Region', 'Product']).agg({
    'Sales': ['sum', 'mean'],
    'Quantity': 'sum',
    'Profit': 'mean'
}).round(2)
```

## 3. Named Aggregations (Cleaner Syntax)

```python
df.groupby('Region').agg(
    total_sales=('Sales', 'sum'),
    avg_sales=('Sales', 'mean'),
    total_qty=('Quantity', 'sum'),
    avg_profit=('Profit', 'mean')
).round(2)
```

## 4. Useful GroupBy Methods

```python
# Size of each group
print("Group sizes:")
print(df.groupby('Region').size())

# First and last records of each group
print("\nFirst record of each Region:")
display(df.groupby('Region').first())

print("\nLast record of each Region:")
display(df.groupby('Region').last())
```

## 5. Filtering Groups

Keep only groups that meet a certain condition.

```python
# Keep only regions where total sales > 700
df.groupby('Region').filter(lambda x: x['Sales'].sum() > 700)
```

## 6. Transform (Keep Original Shape)

`transform` returns a Series/DataFrame with the same index as the original.

```python
# Add a column showing the average sales of that region
df['Region_Avg_Sales'] = df.groupby('Region')['Sales'].transform('mean').round(2)
df
```

# GroupBy and Aggregation Practice

Aaj GroupBy practice kiya. Pandas ka yeh feature kaafi powerful hai data analysis mein.

Jo topics cover kiye:
- Basic groupby
- Multiple aggregations
- Multi-column grouping
- Named aggregations
- transform aur filter

---

```python
import pandas as pd
import numpy as np

# sample sales data banaya practice ke liye
data = {
    'Date': pd.date_range('2024-01-01', periods=12, freq='D'),
    'Region': ['North', 'South', 'East', 'West', 'North', 'South',
               'East', 'West', 'North', 'South', 'East', 'West'],
    'Product': ['A', 'B', 'A', 'C', 'B', 'A', 'C', 'B', 'A', 'C', 'B', 'A'],
    'Sales': [250, 180, 320, 150, 210, 290, 175, 240, 310, 160, 200, 280],
    'Quantity': [10, 7, 12, 5, 8, 11, 6, 9, 13, 4, 7, 10],
    'Profit': [45, 30, 60, 25, 35, 55, 28, 40, 58, 22, 32, 50]
}

df = pd.DataFrame(data)
df
```

## 1. Basic GroupBy

Pehle simple grouping try kiya.

```python
# region wise total sales
df.groupby('Region')['Sales'].sum()
```

```python
# product wise average sales
df.groupby('Product')['Sales'].mean().round(2)
```

```python
# ek saath multiple stats nikalne ke liye
df.groupby('Region')['Sales'].agg(['sum', 'mean', 'count', 'min', 'max'])
```

## 2. Multiple Columns pe GroupBy

```python
# Region aur Product dono se group kiya
df.groupby(['Region', 'Product'])['Sales'].sum().unstack(fill_value=0)
```

```python
# thoda detailed version
df.groupby(['Region', 'Product']).agg({
    'Sales': ['sum', 'mean'],
    'Quantity': 'sum',
    'Profit': 'mean'
}).round(2)
```

## 3. Named Aggregations

Yeh syntax clean lagta hai.

```python
df.groupby('Region').agg(
    total_sales=('Sales', 'sum'),
    avg_sales=('Sales', 'mean'),
    total_qty=('Quantity', 'sum'),
    avg_profit=('Profit', 'mean')
).round(2)
```

## 4. Kuch Useful Methods

```python
# kitne rows hain har group mein
print("Group sizes:")
print(df.groupby('Region').size())

print("\nFirst record of each Region:")
display(df.groupby('Region').first())

print("\nLast record of each Region:")
display(df.groupby('Region').last())
```

## 5. Filtering Groups

Sirf un regions ko rakha jinka total sales 700 se zyada hai.

```python
df.groupby('Region').filter(lambda x: x['Sales'].sum() > 700)
```

## 6. Transform

Original dataframe ka shape same rehta hai transform mein. Yeh kaafi useful hai.

```python
# har row ke saath uske region ka average sales add kar diya
df['Region_Avg_Sales'] = df.groupby('Region')['Sales'].transform('mean').round(2)
df
```

### Q3
Region × Product ke hisaab se total Sales ka table banao.

### Q4
Ek naya column banao jo bataye ki har row ka Sales apne Region ke average se kitna alag hai.

```python
df['Sales_vs_Region_Avg'] = (df['Sales'] - df.groupby('Region')['Sales'].transform('mean')).round(2)
df[['Region', 'Sales', 'Sales_vs_Region_Avg']]
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    "Student": ["Amit", "Neha", "Ravi", "Priya", "Karan"],
    "Branch": ["CSE", "ECE", "CSE", "ECE", "CSE"],
    "Marks": [90, 85, 80, 95, 85]
})
df
```

```python
df.sort_values(by = "Branch")
```

```python
# Break the tie on the basis of the lower marks first
df.sort_values(by = ["Branch","Marks"])
```

```python
# Branch alphabatically asc.
# Marks from highest to lowest within each Branch
```

```python
df.sort_values(by = ["Branch","Marks"],ascending=[True,False])
```

```python
# Sort first on basis of desc order of branch and then on desc order of marks.
df.sort_values(by = ["Branch","Marks"],ascending=[False,False])
```

```python
df = pd.DataFrame({
    "Branch": ["CSE", "CSE", "ECE", "CSE", "ECE", "ECE"],
    "Gender": ["Male", "Female", "Male", "Male", "Female", "Male"],
    "Marks": [85, 90, 80, 85, 95, 80],
    "Student": ["Amit", "Neha", "Ravi", "Karan", "Priya", "Sahil"]
})

df
```

```python
# branch - asc
# gender - desc
# marks - desc
```

```python
df.sort_values(by = ["Branch","Gender","Marks"], ascending=[True,False,False])
```

```python
# a = [2,4a,8,5,4b,3]

# s1 = [2,3,4b,4a,5,8]-->unstable
# s2 = [2,3,4a,4b,5,8]--> stable
```

```python
df = pd.DataFrame({
    "Student":["Amit","Neha","Ravi","Priya"],
    "Marks":[90,85,90,85]
})

# 1-Stable sorting order
# 2-unstable sorting order
```

```python
df.sort_values(by="Marks",kind="stable")
```

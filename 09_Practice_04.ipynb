```python
import numpy as np
import pandas as pd


students = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E"],
    "Maths": [80, np.nan, 70, 90, np.nan],
    "Science": [np.nan, 75, 85, 90, 80],
    "English": [70, 80, np.nan, 95, 85]
})

students
```

```python
# Display the Dataset
students
```

```python
# find the missing value positions.
students.isna()
```

```python
# count missing value in each subject
students[["Maths","Science","English"]].isna().sum()
```

```python
# count all missing cells
students.isna().sum().sum()
```

```python
# Calcualte missing percentage
perc_missing = students.isna().sum().sum()*100/students.size
print(perc_missing)
```

```python
# Display only the names of the students whose maths marks are missing.
df = students["Maths"].isna()
df
```

```python
students[df]["Name"]
```

```python
import pandas as pd

df = pd.DataFrame({
    "Student": ["Amit", "Neha", "Ravi", "Priya", "Karan"],
    "Age": [20, 19, 21, 20, 22],
    "Marks": [85, 92, 76, 95, 88],
    "Attendance": [90, 95, 80, 98, 85]
})

print(df)
```

```python
df
```

```python
# Who has the heighest marks?-->Priya
mx_marks = df["Marks"].max()
mx_marks
```

```python
df[df["Marks"]==mx_marks]["Student"]
```

```python
# Who has the lowest attendence?-->Ravi
lowest_att = min(df["Attendance"])
lowest_att
```

```python
df[df["Attendance"]==lowest_att]["Student"]
```

```python
# Who is the youngest student?-->Neha
youn_stu = min(df["Age"])
youn_stu
```

```python
df[df["Age"]==youn_stu]["Student"]
```

```python
df.sort_values(by = "Age")
```

```python
import pandas as pd

df = pd.DataFrame({
    "Student": ["Amit", "Neha", "Ravi", "Priya", "Karan"],
    "Age": [20, 19, 21, 20, 22],
    "Marks": [85, 92, 76, 95, 88],
    "Attendance": [90, 95, 80, 98, 85]
})

print(df)
```

```python
df
```

```python
df.sort_values(by = "Marks")
```

```python
df_sorted = df.sort_values(by = "Marks")
df_sorted
```

```python
df_sorted = df.sort_values(by = "Marks",ascending=False)
df_sorted
```

```python
# Sort the data to get the most elder person on the top.
df_age = df.sort_values(by = "Age",ascending=False)
df_age
```

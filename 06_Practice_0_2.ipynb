```python
import pandas as pd

students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Karan", "Neha"],
    "Python": [80, 95, 45, 70, 30, 88],
    "SQL": [75, 90, 50, 65, 40, 92],
    "Statistics": [85, 92, 40, 80, 35, 90],
    "Attendance": [90, 95, 65, 80, 55, 98],
    "Branch": ["CSE", "IT", "CSE", "ECE", "ME", "IT"]
})

print(students)
```

```python
# Create a column Academic_year and assign avalue 2026 to each student.
students["Academic_year"] = 2026
students
```

```python
# Assign a mentor to each student.
students["Mentor"] = [
    "Shivam","Riya","Khushi","Kunal","Aman","Shivam"
]
students
```

```python
# Total_marks = python + SQL + STATISTICS
students["Total_marks"] = (students["Python"]
                           +students["SQL"]
                           +students["Statistics"]
                           )
```

```python
students
```

```python
# new_column - % marks
students["Percentage"] =round((
     students["Total_marks"]*100/300),2)
```

```python
students
```

```python
# Average of each subject across all students --> class avg for each subject --> column-wise avg
# Average of the three subjects for each student --> row_wise avg --> avg for each student

# axis = 0 --> column_wise
# axis = 1 --> row_wise

print(students[["Python","SQL","Statistics"]].mean())
```

```python
print(students[["Python","SQL","Statistics"]].mean(axis=1))
```

```python
# Which students scored better in python than in SQL, and by how much?
students["Python_SQL_Difference"] = (
    students["Python"] -students["SQL"]
)
students
```

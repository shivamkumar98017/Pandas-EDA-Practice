```python
# Find the product of all elements except the current element

# Probrlem statement:create a result array in which each position contains
# the product of all the other array elements,without dividing by the current element.

# Example: [1,2,3,4]
# Output:  [24,12,8,6]

# Logic:
# pick an element:
# inner loop: find prd of all, and skip that elem using index.
```

```python
arr = [1,2,3,4]

product = []

for i in range(len(arr)):
  p = 1
  for j in range(len(arr)):
    if i != j:
      p*=arr[j]

  product.append(p)

print(product)
```

```python
# # Left rotate an array by one position

# Problem statement:
# Given an array,all elements one position to the left.
# The first element should move to the last position.

# Example: [10,20,30,40,50]
# OUTPUT:[20,30,40,50,10]
```

```python
a = [10,20,30,40,50]
a[1:]+a[:1]
```

```python
# # Right Rotate an Array by k positions

# Problem statement: Rotate an array to the right by k position.

# Example: [1,2,3,4,5],k=2
# output: [4,5,1,2,3]
```

```python
a = [1,2,3,4,5]
a[3:]+a[:3]
```

```python
#  Find the union of Two arrays

#  problem statement: Point every distinct element that appears in either of two arrays.
#  Do not print duplicate.

# a = [1,2,3,4]
# b = [3,4,5,6]
# ans = [1,2,3,4,5,6]
```

```python
a = [1,2,3,4]
b = [3,4,5,6]

ans = []

for e in a:
  if e not in ans:
    ans.append(e)


for e1 in b:
  if e1 not in ans:
    ans.append(e1)

print(ans)
```

```python
# Find the Intersection without Duplicates

# Problem statement: Print distinct elements common to both arrays,
# even if an element appears multiple times in either innput.

# Example:[2,2,4,6,6,8] and [2,6,,6,10]
# ans = [2,6]
```

```python
a = [2,2,4,6,6,8]
b = [2,6,6,10]

ans = []

for e in a:
  if e in b and e not in ans:
    ans.append(e)

ans
```

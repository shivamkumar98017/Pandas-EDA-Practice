```python
# Find the index where Left and Right products are equal

# Problem statement: Find an index where the product of all elements
# to its left equals the product of all elements to its right.
# Exclude the element at that index from both products.
# Assume the array contains positive integers greater than zero.
# A = [4,3,5,2,6]
# Ans:2
```

```python
a = [4,3,5,2,6]

def product(l):
  p = 1
  for e in l:
    p*=e
  return p

for i in range(len(a)):
  if product(a[:i]) == product(a[i+1:]):


   print(i)
```

```python
# # Answer Multiple Range-Sum Queries

# **Problem statement:** Given an array and several queries, each containing a starting index L and an ending index R,
# calculate the sum of elements from index L to index R inclusive

# **Example:**Array: `[3, 5, 2, 8, 4]`

# - Query L = 1, R = 3 → 15
# - Query L = 0, R = 2 → 10
```

```python
arr = [3, 5, 2, 8, 4]

prefix = [0]
for x in arr:
    prefix.append(prefix[-1] + x)

print(prefix)

def range_sum(L, R):
    return prefix[R + 1] - prefix[L]

print(range_sum(1, 3))
print(range_sum(0, 2))
```

```python
# Calculate the Prefix Sum Array

# **Problem statement:** Given an array, create a new array where each position contains
# the sum of all elements from index 0 to that position.

# **Example:**

# Input: `[2, 4, 1, 5, 3]`

# Output: `[2, 6, 7, 12, 15]`
```

```python
arr = [2,4,1,5,3]
prefix = arr[:1]

for elem in arr[1:]:
  prefix.append(prefix[-1]+elem)

print(prefix)
```

```python
# # Find the Longest Subarray with Equal Numbers of 0s and 1s

# **Problem statement:** Given an array containing only 0 and 1, find the length of the
# longest contiguous subarray containing an equal number of zeros and ones.

# **Example:**

# Input: `[0, 1, 0, 0, 1, 1, 0]`

# Output: `6`
```

```python
array = [0, 1, 0, 0, 1, 1, 0]

longest = 0

for i in range(len(array)):
    zeros = 0
    ones = 0

    for j in range(i, len(array)):
        if array[j] == 0:
            zeros = zeros + 1
        else:
            ones = ones + 1

        if zeros == ones:
            length = j - i + 1
            if length > longest:
                longest = length

print(longest)
```

```python
# Find the Minimum Length Subarray with Sum at Least S

# Problem statement: Given an array of positive integers and a target S,
# find the minimum length of a contiguous subarray whose sum is at least S.
# Print 0 if no such subarray exists.

# Example: [2, 3, 1, 2, 4, 3], S = 7
# output: [4,3] --> 2
```

```python
array = [3,2,1,2,4,3]
s = 7

min_len = 7

for i in range(len(array)):
  total = 0
  for j in range(i,len(array)):
    total = total +array[j]

    if total >= s:
      length = j-i+1
      if length < min_len:
        min_len = length
      break

print(min_len)
```

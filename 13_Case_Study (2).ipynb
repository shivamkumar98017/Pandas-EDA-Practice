```python
import pandas as pd
import numpy as np

data = {
    'Order_ID': [1001, 1002, 1003, 1004, 1005, 1006,
                 1007, 1008, 1009, 1010, 1011, 1001],

    'Customer_Name': ['Amit', 'Priya', 'Rahul', 'Sneha',
                      'Karan', 'Neha', 'Rohit', 'Anjali',
                      'Vikas', 'Pooja', 'Arjun', 'Amit'],

    'Product': ['Laptop', 'Mobile', 'Headphones', 'Tablet',
                'Laptop', 'Mouse', 'Keyboard', 'Mobile',
                'Monitor', 'Tablet', 'Headphones', 'Laptop'],

    'Category': ['Electronics', 'electronics', 'Accessories',
                 'Electronics', 'Electronics', 'accessories',
                 'Accessories', 'Electronics', 'electronics',
                 'Electronics', 'Accessories', 'Electronics'],

    'Quantity': ['1', '2', '1', 'two', '1', '3',
                 '2', '1', '1', '2', '1', '1'],

    'Price': ['55000', '20000', '1500', '18000',
              '52000', '500', '1200', '22000',
              '15000', '17000', '1800', '55000'],

    'Order_Status': ['Delivered', 'delivered', 'Pending',
                     'CANCELLED', 'Delivered', 'pending',
                     'Delivered', 'Shipped', 'delivered',
                     'Pending', 'Delivered', 'Delivered'],

    'Rating': [5, 4, np.nan, 3, 5, 4,
               3, np.nan, 4, 5, 2, 5]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Find orders where Price is greater than 20000.
df[(df["Price"]) > "20000"]
```

```python
# Convert Quantity into numeric formate.
df["Quantity"] = pd.to_numeric(
    df['Quantity'],
    errors='coerce'
)
```

```python
# Convert Price into numeric formate.
df["Price"] = pd.to_numeric(
    df['Price'],
    errors='coerce'
)
```

```python
# Identify records where Quantity could not be converted.
df[df["Quantity"].isnull()]
```

```python
# Repalce invalid Quantity values with the median Quantity.
df["Quantity"] = df['Quantity'].fillna(df["Quantity"].median())
```

```python
# Count missing Ratings.
df["Rating"].isnull().sum()
```

```python
# Calculate the percentage of missing ratings.
perc = df["Rating"].isnull().sum()*100/len(df["Rating"])
perc
```

```python
print(df["Rating"].isnull().mean()*100)
```

```python
# Repalce missing ratings with with the median rating.
df["Rating"] = df["Rating"].fillna(
    df["Rating"].median()
)
df["Rating"].isnull().sum()
```

```python
# Find the duplicate Order_ID records.
df[df.duplicated(subset=["Order_ID"],keep=False)]
```

```python
# Display all duplicate order IDs.
df[df.duplicated(subset=["Order_ID"], keep=False)]["Order_ID"]
```

```python
# Find the number of unique orders.
df["Order_ID"].unique()
```

```python
# Remove duplicate orders while retaining the first record.
df.drop_duplicates(subset="Order_ID", keep="first")
```

```python
# Find Uniqe category Values.
df["Category"].unique()
```

```python
# Convert Category Values to the title case.
df["Category"] = df["Category"].str.strip().str.title()
```

```python
df["Category"].unique()
```

```python
# Standardise order status.
df["Order_Status"] = df["Order_Status"].str.strip().str.title()
```

```python
df
```

```python
# Count Orders under each status.
df["Order_Status"].value_counts()
```

```python
# Create a Total_Amount column using Quantity*Price.
df["Total_Amount"] = df["Quantity"]*df["Price"]
```

```python
df
```

```python
# Classify order as High value if Total_Amount is at least 30,000, otherwise regular.
df["Order_Type"] = np.where(df["Total_Amount"]>=30000,"High Value","Regular")
```

```python
df
```

```python
# Find orders with total amounts greater than 50,000.
df[df["Total_Amount"]>50000]
```

```python
# Rename Customer_Name to Customer.
df.rename(columns={"Customer":"Customer_Name"})
```

```python
# Rename Total_Amount to Order_Value.
df.rename(columns={"Total_Amount":"Order_Value"})
```

```python
# Remove the Rating columns.
df = df.drop(columns=["Rating"])
df
```

```python
import pandas as pd
import numpy as np

data = {
    'Employee_ID': [201, 202, 203, 204, 205, 206,
                    207, 208, 209, 210, 211, 201],

    'Employee_Name': ['Amit', 'Priya', 'Rahul', 'Sneha',
                      'Karan', 'Neha', 'Rohit', 'Anjali',
                      'Vikas', 'Pooja', 'Arjun', 'Amit'],

    'Department': ['IT', 'HR', 'it', 'Finance', 'HR',
                   'IT', 'Finance', 'hr', 'IT', 'Finance',
                   'HR', 'IT'],

    'Salary': ['50000', '45000', '60000', 'invalid',
               '40000', '70000', '55000', '48000',
               '65000', '52000', '42000', '50000'],

    'Attendance': [95, 90, np.nan, 85, 88, 92,
                   75, np.nan, 96, 80, 89, 95],

    'Experience': [2, 4, 6, 1, 3, 8,
                   5, 2, 7, 4, 1, 2],

    'Performance_Rating': [4, 5, 4, 3, np.nan, 5,
                           3, 4, 5, np.nan, 3, 4]
}

df = pd.DataFrame(data)

df
```

```python
# Replace invalid or missing salaries with the median and save the changes in the DataFrame
df['Salary'] = pd.to_numeric(df['Salary'], errors = 'coerce')

df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df["Salary"]
```

```python
# Employee earning more than 50000.
df[df["Salary"]>50000]
```

```python
# Employee having experiance of at least five years.
df[df["Experience"]>=5]
```

```python
# Restore df as a DataFrame and standardize the Department column
df = pd.DataFrame(data)
df["Department"] = df["Department"].str.upper()

# Employees Working in IT.
df[df["Department"] == "IT"]
```

```python
# Employees earning more than ₹45,000 and having at least three years of experiance.
df[(df["Salary"]>"45000")& (df["Experience"]>=3)]
```

```python
# Employees belonging to HR or Finance.
df[df["Department"].isin(["HR", "FINANCE"])]
```

```python
# Convert Salary into numeric formate.
df["Salary"] =pd.to_numeric(df["Salary"],errors="coerce")
```

```python
# Identify invalid salary records.
df[df["Salary"].isnull()]
```

```python
# Filter employees Whose salary is greater than ₹50,000.
df[df["Salary"]>50000]
```

```python
# Convert experiances into interger formate.
df["Experience"] = df["Experience"].astype(int)
```

```python
# Count missing attendence values.
df["Attendance"].isnull().sum()
```

```python
# Replace missing attendence with the average attendence.
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean())
```

```python
df
```

```python
# Replace missing performance rating with the most frequently occurring rating.
df["Performance_Rating"] = df["Performance_Rating"].fillna(df["Performance_Rating"].mode()[0])
```

```python
# Find all records that originally contained missin values.
missing_records = df[df.isna().any(axis=1)]
missing_records
```
